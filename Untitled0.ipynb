```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap

voice_strength = [
    [197, 198, 199, 213, 214, 215, 228, 229, 230, 231, 244, 245, 246, 247],
    [178, 179, 192, 193, 194, 195, 196, 208, 209, 210, 211, 212, 224, 225, 226, 227, 240, 241, 242, 243],
    [128, 129, 130, 131, 144, 145, 146, 147, 160, 161, 162, 163, 176, 177],
    [64, 65, 80, 81, 82, 83, 96, 97, 98, 99, 112, 113, 114, 115],
    [0, 1, 2, 3, 16, 17, 18, 19, 32, 33, 34, 35, 36, 48, 49, 50, 51, 52, 66, 67],
    [4, 5, 6, 7, 20, 21, 22, 23, 37, 38, 39, 53, 54, 55],
    [8, 9, 10, 11, 24, 25, 26, 27, 40, 41, 42, 56, 57, 58],
    [12, 13, 14, 15, 28, 29, 30, 31, 43, 44, 45, 46, 47, 59, 60, 61, 62, 63, 76, 77],
    [78, 79, 92, 93, 94, 95, 108, 109, 110, 111, 124, 125, 126, 127],
    [140, 141, 142, 143, 156, 157, 158, 159, 173, 172, 174, 175, 190, 191],
    [188, 189, 203, 204, 205, 206, 207, 219, 220, 221, 222, 223,
     236, 237, 238, 239, 252, 253, 254, 255],
    [200, 201, 202, 216, 217, 218, 232, 233, 234, 235, 248, 249, 250, 251],
]

# ------------------------------------------------------------------
# Build 16x16 grid
# ------------------------------------------------------------------

grid = np.full((16, 16), -1, dtype=int)

for region, ids in enumerate(voice_strength):
    for idx in ids:
        grid[idx // 16, idx % 16] = region

# ------------------------------------------------------------------
# Better categorical colors
# ------------------------------------------------------------------

cmap = ListedColormap([
    "#d73027",
    "#fc8d59",
    "#fee090",
    "#91bfdb",
    "#4575b4",
    "#66c2a5",
    "#1b9e77",
    "#a6d854",
    "#e5c494",
    "#8dd3c7",
    "#fb8072",
    "#80b1d3",
])

# ------------------------------------------------------------------
# Plot
# ------------------------------------------------------------------

fig, ax = plt.subplots(figsize=(10, 10))

im = ax.imshow(
    grid,
    origin="lower",
    interpolation="nearest",
    cmap=cmap,
)

# cell grid
ax.set_xticks(np.arange(-0.5, 16, 1), minor=True)
ax.set_yticks(np.arange(-0.5, 16, 1), minor=True)
ax.grid(which="minor", color="black", linewidth=0.5)

ax.set_xticks(range(16))
ax.set_yticks(range(16))

# draw region number in every cell
for y in range(16):
    for x in range(16):
        if grid[y, x] >= 0:
            ax.text(
                x,
                y,
                str(grid[y, x]),
                ha="center",
                va="center",
                fontsize=8,
                color="black",
                fontweight="bold",
            )

# thick boundaries between regions
for y in range(16):
    for x in range(16):

        r = grid[y, x]

        if x < 15 and grid[y, x + 1] != r:
            ax.plot(
                [x + 0.5, x + 0.5],
                [y - 0.5, y + 0.5],
                color="white",
                linewidth=3,
            )

        if y < 15 and grid[y + 1, x] != r:
            ax.plot(
                [x - 0.5, x + 0.5],
                [y + 0.5, y + 0.5],
                color="white",
                linewidth=3,
            )

# centroid label for each region
for region in range(12):
    ys, xs = np.where(grid == region)

    if len(xs):
        ax.text(
            xs.mean(),
            ys.mean(),
            f"R{region}",
            fontsize=22,
            fontweight="bold",
            color="white",
            ha="center",
            va="center",
            bbox=dict(
                facecolor="black",
                alpha=0.7,
                boxstyle="round"
            )
        )

plt.title("Voice-strength regions")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------------
# Optional: print sector centroids
# ------------------------------------------------------------------

print("\nRegion centroids:")
for region in range(12):
    ys, xs = np.where(grid == region)
    print(
        f"Region {region:2}: "
        f"x={xs.mean():5.2f}, "
        f"y={ys.mean():5.2f}"
    )
```
